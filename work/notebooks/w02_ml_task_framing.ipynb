## 1. My lane as an ML task (type)

### My Lane: Refresh / Content Opportunity Scoring

This lane is framed as a **ranking/scoring machine learning task**. The objective is to prioritize content pages for manual review using observable search and engagement signals. Since SEO and content teams have limited review capacity, the model ranks pages so reviewers can focus on the highest-priority opportunities for refresh, expansion, protection, pruning, or monitoring.

The system is designed as a **decision-support tool** rather than an automated decision maker. It produces a ranked review queue that helps reviewers allocate their time more effectively based on evidence from the available data.

```python
import pandas as pd
import seaborn as sns

print("ML Task Type : Ranking / Scoring")
print("Decision     : Prioritize content pages for manual review")
print("Output       : Ranked review queue")

df = pd.read_csv('/content_refresh_anonymized (1).csv')

print("Number of content pages:", len(df))

print("\nSample signals available for ranking:")
print([
    "impressions_90d",
    "clicks_90d",
    "ctr",
    "engagement_rate",
    "content_age_days"
])
```

## 2. Target or proxy

The objective is to generate a ranked list of content pages that should be prioritized for manual review. To train this ranking model on the starter dataset, we use a proxy label that identifies whether a page is considered "declining." This label is derived from a defined rule based on the dataset rather than from a future observed outcome. The model learns from observable search and engagement signals (such as impressions, clicks, CTR, engagement metrics, and content metadata) to rank pages according to their review priority.

## 3. Success metric

The primary success metric for this project will be **Precision@20**.

The output of this lane is a ranked review queue. Since reviewers can only inspect a limited number of pages, the most important measure of success is how many of the top-ranked pages are genuinely high-priority review candidates. A higher Precision@20 indicates that the model is effectively identifying the pages that deserve immediate manual attention.

Average Precision can be used as a secondary metric to evaluate the quality of the entire ranking.

```python
top_k = 20

print(f"Chosen evaluation metric: Precision@{top_k}")
print("Secondary metric: Average Precision")
```

## 5. Why ML beats a fixed rule here

A fixed rule-based system would require a large number of manually designed if-else conditions to account for the interactions between search, engagement, and content-related signals. The importance of one feature often depends on the values of several other features. For example, a low CTR may be acceptable for a page with a poor search position but may indicate a problem for a page ranking near the top of search results.

Machine learning can automatically learn these complex relationships from historical data and produce a more flexible and scalable ranking model than manually maintaining hundreds of decision rules.
