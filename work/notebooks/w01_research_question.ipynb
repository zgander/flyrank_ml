## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import seaborn as sns

df = pd.read_csv('/content/content_refresh_anonymized.csv')

print(df.info())

print('\nNumber of pages with high impressions: ',(df['impressions_90d'] > 1000).sum())
print('\nNumber of pages which are declining: ',(df["trend_direction"] == "down").sum())
print('\nSkewness: ',(df['impressions_90d'].skew()))

sns.boxplot(df['impressions_90d'])
```

So, we can make the following inferences from this:
- The starter dataset contains 30,000 entries (or) pages
- 13,509 pages with more than 1000 impressions
- 16,262 pages whose traffic is declining
- Number of pages with a distinct amount of impressions across a wide range (inspite of the high skewness)

These figures suggest that there are enough visible pages and enough performance variation to build and evaluate a meaningful prioritization system.
